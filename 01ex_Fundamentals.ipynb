````python
#excercise 4

def factorial(n):
    if type(n) != int or n < 0:
        print("il tuo numero non è un intero, riprova ")
    elif n == 0 or n == 1:
        return 1
    else:
        #if i don't want to do recursive terms
        #return n * factorial(n-1)
        t = 1
        for i in range(1, n+1):
            t *= i
        return t
x = int(input("inserisci il numero di cui vuoi il fattoriale "))
print(f"il fattoriale è {factorial(x)}")
````
