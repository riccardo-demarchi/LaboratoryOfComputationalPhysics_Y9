```python
import math as m
from functools import reduce

def convert(c, tipo, y): 
    def divisione(numero, dividendo):
        num = []
        h = numero  # deve essere un intero, es. 25
        while h > 0:
            num.append(h % dividendo)  # Resto della divisione
            h = h // dividendo         # Divisione intera

        return num
    cont = {}
    cont = dict(zip(y, [None]*len(y)))
    c_str = str(c).strip()
    
    let = ['A', 'B', 'C', 'D', 'E', 'F']
    num = [10, 11, 12, 13, 14, 15]
    if tipo == 'binary':
        b = list(map(int, c_str))
        strbin = "".join(str(x) for x in b)
        sign = (-1) ** b[0]
        cont[y[1]] = strbin
            
        count = 0
        
        #so now the dict is {dec:, bin:, hex:}
        for i in range(1,len(b)):
            count += b[i]*2**(len(b)-1-i)

        cont[y[0]] = count*sign
            #from binary to dec
            #how to associate cont[0] to decimal ?

        del b[0]
        if len(b) % 4 != 0:
            while len(b) % 4 != 0:
                b.insert(0, 0)
        #number of chunks 4
        li = []
        li = [b[i:i + 4] for i in range(0, len(b), 4)]
        
        newv = []
        for o in range(len(li)):
            cut = 0
            for j in range(4):
                cut += li[o][j]*2**(4-1-j)
            newv.append(cut)
        for o in range(len(newv)):
            for k in range(len(num)):
                if newv[o] == num[k]:
                    newv[o] = let[k]
                else: 
                    continue
            #every 4 steps save four step of the list
            #from binary to hex

        if sign > 0:
            newv.insert(0, '+')
        else:
            newv.insert(0, '-')

        # Convert each element to string, then join
        result = "".join(str(x) for x in newv)        
        cont[y[2]] = result
        
        
        return cont

        
    elif tipo == 'decimal':
        # Handles numbers like "-23" or "+23"
        if c_str.startswith('-'):
            sign_char = '-'
            valore_assoluto = int(c_str[1:])
        elif c_str.startswith('+'):
            sign_char = '+'
            valore_assoluto = int(c_str[1:])
        else:
            sign_char = '+'
            valore_assoluto = int(c_str)
            
        cont[y[0]] = c_str
        #dec fatto 

        #dec to bin
        
        
        newv = divisione(valore_assoluto, 2)

        newv.reverse()
        if sign_char == '+':
            newv.insert(0, 0)
        else:
            newv.insert(0, 1)

        # Convert each element to string, then join
        result = "".join(str(x) for x in newv)        
        cont[y[1]] = result

        #now hexadecimal
        new2 = divisione(valore_assoluto, 16)
        new2.reverse()

        for o in range(len(new2)):
            for k in range(len(num)):
                if new2[o] == num[k]:
                    new2[o] = let[k]
                else: 
                    continue
        new2.insert(0, sign_char)

        result2 = "".join(str(x) for x in new2)        
        cont[y[2]] = result2


        return cont

    else:
        # Gestisce input esadecimali come "-1A", "0AB12", "+3E"
        c_str_upper = c_str.upper()

        if c_str_upper.startswith("-"):
            sign_char = "-"
            hex_body = c_str_upper[1:]
        elif c_str_upper.startswith("+"):
            sign_char = "+"
            hex_body = c_str_upper[1:]
        else:
            sign_char = "+"
            hex_body = c_str_upper
        
        cont[y[2]] = c_str_upper
        
        count = 0
        n = len(hex_body)
        #so now the dict is {dec:, bin:, hex:}
        for i in range(n):
            char = hex_body[i]
            if char in let:
                val = num[let.index(char)]
            else:
                val = int(char)
            count += val * (16 ** (n - 1 - i))

        if sign_char == "-":
            cont[y[0]] = -count
        else:
            cont[y[0]] = count
        #for i in range(1,len(b)):
        #    cont[0] += b[i]*16**(len(b)-1-i
        
        
        if count == 0:
            bits = [0]
        else:
            bits = divisione(count, 2)
            bits.reverse()

        sign_bit = 1 if sign_char == "-" else 0
        bits.insert(0, sign_bit)
        
        cont[y[1]] = "".join(str(x) for x in bits)

        return cont

y = ['decimal', 'binary', 'hexadecimal']
dec = ""  # Inizializzazione della variabile

# Continua a chiedere finché l'input NON fa parte della lista y
while dec not in y:
    dec = input("In which basis are you going to put your number (hexadecimal, decimal, binary) ? ").strip().lower()

# Ora che dec è valido, chiedi il numero
number = input("What's your number ? ")

res = convert(number, dec, y)
print(res)
 
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
#exercise 2 

def binary_to_floating(x):
    b = list(map(int, str(x)))
    sign = (-1)**b[0]

    count = 0
    for i in range(1, 9):
        #so now the dict is {dec:, bin:, hex:}
        count += b[i]*2**(8-i) 
    e = count - 127

    #mantissa
    m = 0
    for i in range(9, len(b)):
        m += b[i]*2**(8-i)
    M = m + 1.0

    return sign*M*2**e
        





inp = input("write 32 bit word to convert ")

result = binary_to_floating(inp)

print(f"the number in floating is {result}")

```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
import math as m
import sys

def over():
    val = 1.0
    while val*2 != float("inf"):
        val *= 2
    return val

def under():
    val = 1.0
    while val/2 != 0.0:
        val /= 2
    return val    

def main():
    overflow = over()
    underflow = under()

    print(f"Overflow limit (within factor of 2):  {overflow}")
    print(f"Underflow limit (within factor of 2): {underflow}")
    

if __name__ == "__main__":
    main()
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
#excercise 7
import numpy as np
import math as m
import timeit


def n_steps(t, n):
    return n//t   

def integral(x):
    I = 0
    for i in range(len(x)):
        I += h*(m.sqrt(1-x[i]**2))
    return I 

#a
N = 100
h = 2/N

x = np.arange(-1, 1, h)

result1 = integral(x)

print(f"The numerical result for semi circle area is {result1}")
print(f"the real area value is {m.pi/2}")


tempo = timeit.timeit(lambda: integral(x), number = 1)
result_n = n_steps(tempo, N)

print(f"the maximum N that can be increased until the time of compute is more than 1s is {result_n}")
```
