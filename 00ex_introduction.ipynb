```python
#Excercise 3
import numpy as np
import math

def dist(u, v):
    x = []
    c = 0
    for i in range(len(u)): 
        x.append((u[i]-v[i])**2)
    for i in range(len(x)):
        c += x[i]
    return np.sqrt(c)

def main():

    n = int(input("inserisci la dimensione dove vuoi lavorare "))
    u = []
    v= []
    for i in range(n):
        valore = int(input(f"inserisci la coordinata x-{i+1}esima del vettore u: "))
        u.append(valore)

    for i in range(n):
        valore = int(input(f"inserisci la coordinata x-{i+1}esima del vettore v: "))
        v.append(valore)
    

    distanza = dist(u, v)

    print(f"La distanza tra i due vettori è : {distanza}")

    



if __name__ == "__main__":
    main()
```
